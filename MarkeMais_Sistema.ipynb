```python
#Loja MarkMais
#Parte 1- Listas

print("======== Bem-vindo à Loja MarkMais! ========")
def sistema_markmais():

# Parte 1: Listas Iniciais da MarkMais
    produtos = ["Camiseta Básica", "Calça Jeans", "Jaqueta Corta-Vento"]
    precos = [49.90, 129.90, 219.90]

# Histórico para o relatório final
    inseridos_historico = []
    apagados_historico = []

    while True:
        print("\n" + "="*40)
        print("ESTOQUE ATUAL - MARKMAIS")
        print("="*40)
        if len(produtos) > 0:
            for p, pr in zip(produtos, precos):
                print(f"- {p}: R$ {pr:.2f}")
        else:
            print("Estoque vazio!")
        print("="*40)

# Menu de Ações
        print("\nO que você deseja fazer?")
        print("[1] Adicionar novo produto e valor")
        print("[2] Excluir o último item do estoque")
        print("[3] Excluir um item específico pelo nome")
        print("[4] Finalizar e ver relatório completo")

        opcao = input("\nEscolha uma opção (1, 2, 3 ou 4): ")

        if opcao == "1":

# Parte 2: Adicionar produtos e valores
            novo_produto = input("Digite o nome da nova peça de roupa:")
            novo_preco = float(input("Digite o preço da nova peça: "))

            produtos.append(novo_produto)
            precos.append(novo_preco)
            inseridos_historico.append((novo_produto, novo_preco))

            print(f"\n[Sucesso] '{novo_produto}' adicionado ao estoque!")

        elif opcao == "2":

# Parte 3: Excluir o último item
            if len(produtos) > 0:
                produto_removido = produtos.pop()
                preco_removido = precos.pop()
                apagados_historico.append((produto_removido, preco_removido))
                print(f"\n[Removido] Último item apagado: {produto_removido} \
                 (R$ {preco_removido:.2f})")
            else:
                print("\n[Aviso] O estoque já está vazio, não há o que apagar!")

        elif opcao == "3":

# Parte 4: Excluir por nome específico
            item_remover = input("Digite o nome exato da peça que deseja excluir:")
            if item_remover in produtos:
                i = produtos.index(item_remover)
                prod_rem = produtos.pop(i)
                preco_rem = precos.pop(i)
                apagados_historico.append((prod_rem, preco_rem))
                print(f"\n[Removido] O item '{prod_rem}' (R$ {preco_rem:.2f}) \
                foi excluído com sucesso.")
            else:
                print("\n[Erro] Esse item não existe no estoque!")

        elif opcao == "4":

#Relatório Final e Opção de Reiniciar
            print("\n" + "="*40)
            print("RELATÓRIO FINAL MARKMAIS")
            print("="*40)
            print("📦 Estoque Atual nas Prateleiras:")
            if len(produtos) > 0:
                for p, pr in zip(produtos, precos):
                    print(f"  - {p}: R$ {pr:.2f}")
            else:
                print("  - Nenhum produto no estoque.")

            print("\n➕ O que foi inserido durante a sessão:")
            if inseridos_historico:
                for p, pr in inseridos_historico:
                    print(f"  - {p} (R$ {pr:.2f})")
            else:
                print("  - Nenhum item novo foi inserido.")

            print("\n❌ O que foi apagado durante a sessão:")
            if apagados_historico:
                for p, pr in apagados_historico:
                    print(f"  - {p} (R$ {pr:.2f})")
            else:
                print("  - Nenhum item foi apagado.")
            print("="*40)

# Opção de reiniciar ou finalizar de vez
            reiniciar = input("\nDeseja reiniciar a operação ou finalizar? \
            (Digite 'reiniciar' ou 'finalizar'): ").strip().lower()
            if reiniciar == 'reiniciar':

# Reinicia as listas e históricos se desejar começar do zero
                produtos = ["Camiseta Básica", "Calça Jeans", "Jaqueta Corta-Vento"]
                precos = [49.90, 129.90, 219.90]
                inseridos_historico.clear()
                apagados_historico.clear()
                print("\n[Sistema reiniciado com sucesso!]")
            else:
                print("\nOperação finalizada. Até logo!")
                break
        else:
            print("\n[Opção Inválida] Por favor, escolha entre 1, 2, 3 ou 4.")

# Executa o sistema
sistema_markmais()
```
